# Wrist Degrees of Freedom (Project Standard)

| Wrist DOF | Rotation Axis | Primary Visual Effect |
|-----------|---------------|-----------------------|
| **Wrist Roll** | Around the forearm's longitudinal axis | Rotates the palm between downward- and upward-facing orientations (pronation/supination). |
| **Wrist Pitch** | Across the wrist hinge | Tilts the hand upward or downward (flexion/extension). |
| **Wrist Yaw** | Vertical axis through the wrist | Swings the hand left or right relative to the forearm. |

```python
LEFT_CUBE_DISPLAY = [

    # ---------------- LEFT ARM ----------------

    0.00,   # LeftShoulderPitch
            # Raises/lowers the arm in the sagittal plane.

    1.20,   # LeftShoulderRoll
            # Moves the arm laterally away from the torso.

    0.00,   # LeftShoulderYaw
            # Rotates the upper arm about its long axis.

    0.80,   # LeftElbow
            # Bends the elbow to support the cube.

    # ---------------- LEFT WRIST ----------------

    -3.00,  # LeftWristRoll
            # Rotation about the forearm axis.
            # Primary contributor to the palm-up presentation.
            # Produces pronation/supination.

    0.00,   # LeftWristPitch
            # Wrist hinge motion.
            # Tilts the hand up/down.
            # Produced little visible effect in this experiment.

    0.75,   # LeftWristYaw
            # Swings the hand left/right.
            # Fine adjustment used to improve cube visibility.

    # ---------------- RIGHT ARM ----------------

    0.00,   # RightShoulderPitch
   -1.57,   # RightShoulderRoll
    0.00,   # RightShoulderYaw
    1.57,   # RightElbow
    0.00,   # RightWristRoll
    0.00,   # RightWristPitch
    0.00    # RightWristYaw
]
```

### Experimental observations

- A large negative **LeftWristRoll** was the dominant factor in rotating the palm upward.
- **LeftWristYaw** provided a small but noticeable improvement to the display angle.
- **LeftWristPitch** had little or no visible contribution in this specific arm configuration.
- This pose successfully displayed the foam cube to the audience while maintaining the grasp.
