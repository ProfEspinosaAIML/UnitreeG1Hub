- This notebook captures the best current understanding of the hand model derived from experimentation.


## DDS Topics

Command topic:
`rt/dex3/right/cmd`

State topic:
`rt/lf/dex3/right/state`


## Current 7-Channel Hand Model

Experimental results indicate:

| Channel | Status | Meaning |
|----------|----------|----------|
| q0 | Verified | Thumb rotation |
| q1 | Unknown / inactive | No observed motion |
| q2 | Unknown / inactive | No observed motion |
| q3 | Verified | Right finger actuator |
| q4 | Verified | Right finger actuator |
| q5 | Verified | Left finger actuator |
| q6 | Verified | Left finger actuator |

When facing the palm of the hand:

- Right finger = channels q3 + q4
- Left finger = channels q5 + q6
- Thumb rotation = q0


## Known Limitations

- Thumb rotation is verified.
- Thumb curl has not been discovered.
- Channels q1 and q2 have not produced observable motion.
- Extensive excitation testing did not reveal additional thumb DOFs.


## Relationship to Demo v0 (`g1_fingers_sdk_dds_demo1.py`)

Demo v0 is built entirely from verified DOFs:

1. Right finger close/open
2. Left finger close/open
3. Alternating finger motions
4. Thumb rotation showcase
5. Grasp illusion using both fingers and thumb rotation
6. Dexterity finale

No q1/q2 commands are used in the demo.
