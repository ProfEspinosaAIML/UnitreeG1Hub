# Coordinate Model Used By Demo v2

The demo simplifies control into three logical degrees of freedom:

```python
thumb_rotation = q0
right_finger   = q3 + q4
left_finger    = q5 + q6
```

Logical pose:

```python
pose = [
    q0,
    0.0,
    0.0,
    right,
    right,
    left,
    left,
]
```

Assumption used throughout Demo v2:

```python
0.0 = open
1.0 = closed
```



# Mapping Between Code And Motion

## Open Hand

```python
move_pose(
    q0=0,
    right=0,
    left=0
)
```

Physical result:

- Thumb centered
- Right finger extended
- Left finger extended

---

## Close Right Finger

```python
move_pose(
    right=1
)
```

Physical result:

- Right finger closes
- Left finger remains open
- Thumb unchanged

---

## Close Left Finger

```python
move_pose(
    left=1
)
```

Physical result:

- Left finger closes
- Right finger remains open
- Thumb unchanged



# Thumb Rotation

Clockwise rotation:

```python
move_pose(
    q0=1
)
```

Counter-clockwise rotation:

```python
move_pose(
    q0=-1
)
```

Observed behavior:

- Thumb rotates approximately 135–140 degrees total travel.
- No thumb curl detected.
- Rotation works reliably and repeatably.



# Demo v2 Choreography

## Scene 1 — Activation

Purpose:

Introduce thumb motion.

Sequence:

1. Open hand
2. Thumb clockwise
3. Center
4. Thumb counter-clockwise
5. Center

---

## Scene 2 — Finger Discovery

Purpose:

Show independent finger control.

Sequence:

1. Right finger close/open
2. Right finger close/open
3. Left finger close/open
4. Left finger close/open

---

## Scene 3 — Dexterity Exercise

Purpose:

Show coordination between both fingers.

Sequence:

Right → Left → Both → Open

Repeated three times.



# Demo v2 Choreography (continued)

## Scene 4 — Pinch Illusions

Purpose:

Create the appearance of grasping despite lack of thumb curl.

Techniques:

- Rotate thumb.
- Close one finger.
- Hold briefly.
- Reopen.

This produces visually convincing pinch-like motions.

---

## Scene 5 — Power Grasp

Purpose:

Create strongest grasp illusion currently possible.

Pose:

```python
q0=-1
right=1
left=1
```

This combines:

- inward thumb rotation
- both fingers closed

Although not a true grasp, it appears grasp-like from many viewing angles.



# Demo v2 Choreography (continued)

## Scene 6 — Finger Wave

Purpose:

Demonstrate rapid alternation.

Pattern:

```text
Right
Both
Left
Open
```

followed by the reverse ordering.

---

## Scene 7 — Finale

Purpose:

Combine all verified motions.

Features:

- Alternating finger taps
- Simultaneous finger closure
- Thumb sweep
- Return to fully open hand
