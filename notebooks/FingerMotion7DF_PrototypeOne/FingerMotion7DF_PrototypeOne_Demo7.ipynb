Palm facing observer:

```text
Finger A     Thumb     Finger B
    |          |           |
   q3/q4      q0         q5/q6
```

## Left Hand Kinematic Map

| Channel | Function |
|----------|----------|
| q0 | Thumb rotation |
| q1 | Inactive / firmware reserved |
| q2 | Inactive / firmware reserved |
| q3 | Finger A base joint |
| q4 | Finger A tip joint |
| q5 | Finger B base joint |
| q6 | Finger B tip joint |

Where:

- Finger A = finger left of thumb when viewing palm
- Finger B = finger right of thumb when viewing palm
