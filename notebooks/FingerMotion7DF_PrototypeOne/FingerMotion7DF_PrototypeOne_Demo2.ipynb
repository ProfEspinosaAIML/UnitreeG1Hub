# FingerMotion7DF - PrototypeOne - Demo2

Documentation for Demo v1 (`g1_fingers_sdk_dds_demo2.py`) and the video `g1_fingers_sdk_dds_demo2.mp4`.

# Current Best Hand Model

State vector: `[q0,q1,q2,q3,q4,q5,q6]`

- q0 = thumb rotation
- q1 = unknown
- q2 = unknown
- q3+q4 = RIGHT finger
- q5+q6 = LEFT finger

Practical convention:
- 0.0 = open
- 1.0 = closed

# Experimental Findings

Thumb rotation is verified.
No verified thumb curl has been discovered.
Channels q1 and q2 produced no repeatable observable motion in testing.

# Mapping to Demo v1

`move_pose(q0,right,left,hold)`

Examples:

Open hand:
`move_pose(0,0,0)`

Right finger close:
`move_pose(0,1,0)`

Left finger close:
`move_pose(0,0,1)`

Both fingers close:
`move_pose(0,1,1)`

Thumb clockwise:
`move_pose(1,0,0)`

Thumb counter-clockwise:
`move_pose(-1,0,0)`
