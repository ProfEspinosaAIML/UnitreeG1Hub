# 10. Main Execution (Expanded)

Entry point: initializes system and runs controller.

---

```python id="m2x8k4"
if __name__ == '__main__':
    print("WARNING: Ensure robot is safely supported (gantry attached).")
    input("Press Enter to start...")

    if len(sys.argv) > 1:
        ChannelFactoryInitialize(0, sys.argv[1])
    else:
        ChannelFactoryInitialize(0)

    ctrl = LeftArmRaise()
    ctrl.Init()
    ctrl.Start()

    while True:
        time.sleep(1)
        if ctrl.done:
            sys.exit(0)
```

---

## 🧠 Big Picture: What This Section Does

This block defines the **program lifecycle**:

```text
Initialize → Connect → Start Control → Monitor → Exit
```

---

```python
if __name__ == '__main__':
```

---

### What does this mean?

This ensures:

```text
Code runs ONLY when script is executed directly
```

---

### Not when:

* Imported as a module
* Used in another script

### 🧠 Why this matters

This is standard Python practice for:

* Reusability
* Modular design

## ⚠️ Step 2: Safety Warning

```python
print("WARNING: Ensure robot is safely supported (gantry attached).")
input("Press Enter to start...")
```

---

### Why is this critical?

Because:

```text
You are about to command a real humanoid robot
```

---

### Risks:

* Unexpected motion
* Loss of balance
* Hardware damage

---

### 🧠 Engineering Insight

This is a **human-in-the-loop safety barrier**.

---

### The `input()` call:

* Pauses execution
* Forces operator confirmation
* Ensures readiness

## 🌐 Step 3: Initialize DDS Communication

```python
if len(sys.argv) > 1:
    ChannelFactoryInitialize(0, sys.argv[1])
else:
    ChannelFactoryInitialize(0)
```

---

### What does this do?

Initializes the **DDS middleware** used to communicate with the robot.

---

### Case 1: Interface specified

```bash
python3 script.py eth0
```

→ Uses:

```python
ChannelFactoryInitialize(0, "eth0")
```

---

### Case 2: Default interface

```python
ChannelFactoryInitialize(0)
```

---

### 🧠 Why this matters

The robot is connected via:

```text
Ethernet (e.g., eth0)
```

---

### If wrong interface is used:

* ❌ No communication
* ❌ No state updates
* ❌ No motion

---

### ⚠️ Common failure mode

```text
Script runs → but robot does nothing
```

Cause:

```text
DDS not bound to correct network interface
```

```python
ctrl = LeftArmRaise()
```

---

### What happens here?

  * control parameters
  * joint lists
  * target trajectories

## 📡 Step 5: Initialize Communication Channels

```python
ctrl.Init()
```

---

### This sets up:

* Publisher (`rt/arm_sdk`)
* Subscriber (`rt/lowstate`)

### Without this:

```text
Controller cannot talk to robot
```

---

## ▶️ Step 6: Start Control Loop

```python
ctrl.Start()
```

---

### This does:

1. Waits for first state update
2. Starts real-time control thread

---

### After this line:

```text
Robot control is ACTIVE
```

---

### 🧠 Important

This is where:

```text
Program transitions from setup → execution
```

---

## 🔄 Step 7: Main Monitoring Loop

```python
while True:
    time.sleep(1)
    if ctrl.done:
        sys.exit(0)
```

---

### What is this loop doing?

It is **not controlling the robot**.

Instead, it:

```text
Monitors the controller state
```

---

### Why `sleep(1)`?

* Reduces CPU usage
* No need for high-frequency checks

---

### 🧠 Separation of concerns

| Component          | Role              |
| ------------------ | ----------------- |
| ControlLoop thread | real-time control |
| Main loop          | supervision       |

---

## 🏁 Step 8: Exit Condition

```python
if ctrl.done:
    sys.exit(0)
```

---

### Where does `ctrl.done` come from?

Set in **Final Stage (9.6)**:

```python
self.done = True
```

---

### Meaning:

```text
Motion complete → exit program cleanly
```
